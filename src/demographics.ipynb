### Education level

```python
edu_current_df["Percentage"] = edu_current_df["n"] / 319 * 100
edu_current_df
```

```python
edu_finished_df["Percentage"] = edu_finished_df["n"] / 319 * 100
edu_finished_df
```

```python
edu_total_df = pd.merge(edu_current_df, edu_finished_df, left_on="education1a_t1", right_on="education1b_t1", suffixes=("_current", "_finished"), how="outer").fillna(0)
edu_total_df["Total"] = edu_total_df["n_current"] + edu_total_df["n_finished"]
edu_total_df["Percentage"] = edu_total_df["Total"] / 319 * 100
edu_total_df["Education level"] = edu_total_df["education1b_t1"]
edu_total_df = edu_total_df[["Education level", "Total", "Percentage"]]

primary = ["De basisschool"]
low = ["VMBO", "MBO"]
medium = ["HBO", "HAVO"]
high = ["VWO/gymnasium", "WO"]

education_levels = {
    "Primary": primary,
    "Low": low,
    "Medium": medium,
    "High": high
}

for level, categories in education_levels.items():
    total = edu_total_df[edu_total_df["Education level"].isin(categories)]["Total"].sum()
    percentage = edu_total_df[edu_total_df["Education level"].isin(categories)]["Percentage"].sum()
    print(f"{level} & {int(total)} ({percentage:.1f}\%) \\\\")
```

### Gender composition
